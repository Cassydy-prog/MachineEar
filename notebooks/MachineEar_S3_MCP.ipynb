# MachineEar S3 — MCP
Use a local kernel with this repository as the current directory. No API key needed.
Discovery tests the real transport; real search requires the local embedding runtime.

```python
import json
import os
import subprocess
import sys

async def query_server(query=None, machine_type="fan"):
    command = [sys.executable, "-m", "agentic.mcp_client"]
    if query is not None:
        command += ["--query", query, "--machine-type", machine_type]

    result = subprocess.run(
        command,
        capture_output=True,
        encoding="utf-8",
        env={**os.environ, "PYTHONIOENCODING": "utf-8"},
        timeout=150,
    )
    if result.returncode != 0:
        raise RuntimeError(result.stderr.strip() or result.stdout.strip())

    response = json.loads(result.stdout)
    if isinstance(response, dict) and response.get("isError"):
        raise RuntimeError(response)
    return response

await query_server()
```

```python
# Real retrieval: enable only after preparing local embeddings.
RUN_REAL_SEARCH = True
if RUN_REAL_SEARCH:
    result = await query_server("fan bearing lubrication", "fan")
    print(result)
    assert not result.get("isError"), "MCP returned an error; do not count this as a pass"
```
